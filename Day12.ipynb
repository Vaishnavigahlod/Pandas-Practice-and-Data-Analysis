```python
import pandas as pd
data = {
    "Patient_ID": [101,102,103,104,105,106,107,108,109,110],
    "Age": [25,45,35,60,50,28,65,40,55,30],
    "Gender": ["Male","Female","Male","Female","Male",
               "Female","Male","Female","Male","Female"],
    "Department": ["Cardiology","Neurology","Cardiology","Orthopedic",
                   "Neurology","Cardiology","Orthopedic","Neurology",
                   "Cardiology","Orthopedic"],
    "City": ["Nagpur","Pune","Nagpur","Mumbai","Pune",
             "Nagpur","Mumbai","Pune","Nagpur","Mumbai"],
    "Treatment_Cost": [50000,35000,60000,45000,30000,
                       55000,70000,40000,65000,50000],
    "Hospital_Stay": [5,3,6,4,2,5,7,3,6,4]
}

df = pd.DataFrame(data)

```

```python
print(df)
```

```python
#duplicate value banana, we use concat function:to merge 2 data frame , humari rows 10 se 12 ho gai duplicate 
#records aa gaye iss liye, ignore index true matlab purana index bhul jaao new index se dekho.
df_duplicate =pd.concat([df,df.iloc[[2,5]]],ignore_index=True)
df_duplicate #print kiya 
```

pd.concat(): dataframes ko jodne ka kaam karta hai
iloc[[2,5]] row 2 our row 5 select karna
ignore_index = True--->index ko fresh nummbering denga new no

```python
#agar dataframe me 10 rows hai or hum df.iloc[[2,5]]se 2 rows ko or add kar rhe hai toh total kitni row hongi
#ans:12 rows

#aur konsi row duplicate hongi?
#ans:patient_id-- 103,106
```

```python
#imagin karo hospital ke database me patient 103 ka same record do baar aa gya. ab hospital ke total patients
#count karenge to kya honga?
#use duplicated function(): duplicate hai ya nhi ye janne ke liye ye use karte hain.
df_duplicate.duplicated()
```

```python
#sirf duplicate row dekhne ke liye
df_duplicate[df_duplicate.duplicated()]
```

```python
#duplicate record hatane ke liye
df_clean = df_duplicate.drop_duplicates()
df_clean #purana data frame vapas aa gya
```

```python
df_clean.shape #kitni rows hai dekhne ke liye
```

duplicated() :identify duplicate values
drop_duplicates() :duplicate ko remove karta hai

```python
#patient 103 do baar hai to kya dono record delete honge kya ------>NO


#patient id same hai but tratment cost alag hai to duplicate honga kya ---->NO
df_duplicate.duplicated(subset=["Patient_ID"]) #duplicate value true ho jayngi
```

```python
#abb duplicate value hatane ke liye
df_duplicate.drop_duplicates(subset=["Patient_ID"])
```
