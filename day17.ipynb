```python
import pandas as pd
dates =pd.Series(pd.to_datetime(["2016-12-23","2020-01-15","2023-08-05"]))
print(dates.dt.year)
print(dates.dt.month)
print(dates.dt.day)
```

```python
#month name
dates.dt.month_name()
```

```python
#name of weekday
dates.dt.day_name()
```

```python
import pandas as pd
df = pd.read_csv(r"D:\data analyst\pandas\DATA\DataFrame\movies.csv")
df
```

```python
print(df[["title_x","release_date"]])
```

```python
#now convert the data from original data to Nat
df["release_date"]=pd.to_datetime(df["release_date"],format="mixed",errors="coerce")
```

```python
df["release_year"] = df["release_date"].dt.year
df["release_month"] = df["release_date"].dt.month
df["release_month_name"] = df["release_date"].dt.month_name()
df["release_day"] = df["release_date"].dt.day
df["release_weekday"] = df["release_date"].dt.day_name()

print(df[[
    "title_x",
    "release_date",
    "release_year",
    "release_month_name",
    "release_weekday"
]].head(10))
```
