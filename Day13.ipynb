GROUPBY =it groups rows with the same values together, so you can do calculations on each group.
        (same name walo ko ek group me daal deta hai fir us group pe calculation kar sakte hai)

```python
#groupby: 
#genre-->wise avg rating kya hai
#
csv_movies.groupby("Genre")["IMDB_Rating"].mean()
```

```python
#genre- wise avg movies run time
csv_movies.groupby("Genre")["Runtime"].mean()
```

```python
#genre-wise total number of votes using groupby
csv_movies.groupby("Genre")["No_of_Votes"].sum() 
```

```python
#genre-wisemax imdb rating
csv_movies.groupby("Genre")["IMDB_Rating"].max()
```

Agrigation function(agg)= it takes many values and returns a single summary value.(bohot sare values ko jod ke ek 
                           single value banata hai)

```python
#genre wise avg rating cahiye, max rating, min rating, avg runtime ek report(agg-agrigation function)
csv_movies.groupby("Genre").agg(Avg_Rating =("IMDB_Rating","mean"),
Max_Rating=(("IMDB_Rating","min")),
Avg_Runtime=(("Runtime","mean")))
                                
```
