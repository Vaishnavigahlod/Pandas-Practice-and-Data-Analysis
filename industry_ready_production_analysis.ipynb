```python
#problem--1)Audiance engagement
#which movie have the strongest audiance engagement signal in this data set
csv_movies.nlargest(10,"No_of_Votes")[["Series_Title","Genre","IMDB_Rating","No_of_Votes"]]
```

```python
#problem--2)High Quality Movies Shortlist
#find movies with rating at least  8.5 and at least 100,000
shortlist=csv_movies[(csv_movies["IMDB_Rating"]>=8.5)&(csv_movies["No_of_Votes"]>=100000)]
shortlist[["Series_Title","IMDB_Rating","No_of_Votes"]]
```

```python
#problem--3) Genre performnce report
#which genre have high ratings, longer runtimes and more audiance votes
#(logic,gpyb,agg,max,mean,sum)


genre_report = csv_movies.groupby("Genre").agg(
    Movie_Count=("Series_Title", "count"),
    Average_Rating=("IMDB_Rating", "mean"),
    Median_Rating=("IMDB_Rating", "median"),
    Average_Runtime=("Runtime", "mean"),
    Total_Votes=("No_of_Votes", "sum")
)

genre_report.sort_values(
    "Average_Rating", ascending=False
)
```
