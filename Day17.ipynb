(question)Recipe website ki recipe list do alag parts mein mili hai. Manager ko dono lists ko ek list mein combine karna hai
Do DataFrames hain. Kya humein 
rows jodni hain ya columns?
Dono DataFrames mein same columns hone chahiye ya different?
Agar pehli list mein 5 rows aur doosri mein 5 rows hain, toh combine karne par kitni rows hongi?

```python
#(answer):
first_list = recipes[["recipe_name", "rating"]].iloc[:5]
second_list = recipes[["recipe_name","rating"]].iloc[5:10]

print(first_list)
print(second_list)

combined = pd.concat([first_list, second_list],ignore_index=True)

print(combined)
print(combined.shape)
```
