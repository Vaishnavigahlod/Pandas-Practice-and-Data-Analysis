```python
import pandas as pd

data = {
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "Age": [25, 45, None, 60, 50, 28, 65, None, 55, 30],

    "Gender": [
        "Male", "Female", "Male", "Female", None,
        "Female", "Male", "Female", "Male", "Female"
    ],

    "Department": [
        "Cardiology", "Neurology", "Cardiology", "Orthopedic",
        "Neurology", "Cardiology", "Orthopedic", "Neurology",
        "Cardiology", "Orthopedic"
    ],

    "City": [
        "Nagpur", "Pune", "Nagpur", None,
        "Pune", "Nagpur", "Mumbai", "Pune",
        "Nagpur", "Mumbai"
    ],

    "Treatment_Cost": [
        50000, 35000, None, 45000, 30000,
        55000, 70000, 40000, None, 50000
    ],

    "Hospital_Stay": [
        5, 3, 6, None, 2,
        5, 7, 3, 6, 4
    ]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df.isnull() #jaha value nhi thi vaha true aa gya 
```

```python
df.isnull().sum()
```

DROPNA() :dropna() is used to remove missing values (NaN/ Null) from dataframe(for remove row or column)
          (missing record hatana)

```python
df.dropna() #it remove whole all row
```

```python
df.dropna(subset = ["Age"]) #sirf age vali jo row nan thi vo chale gai
```

FILLNA() :IT is used to fill or replace missing values (NaN/Null) with some other value(missing value bharo)

```python
df["Age"].fillna(40)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Age"]
```
