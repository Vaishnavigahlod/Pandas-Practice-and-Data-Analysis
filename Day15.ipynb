```python
import pandas as pd

movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy"]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(movies)
```

```python
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(ratings)
```

```python
#left join
print(movies.join(ratings,how='left'))
```

```python
#right join (right table index)
print(movies.join(ratings,how='right'))     
```

```python
# inner (it show only matching index)
print(movies.join(ratings,how="inner"))
```

```python
#outer (dono table ke sabhi index show karenga)
print(movies.join(ratings,how='outer'))
```
